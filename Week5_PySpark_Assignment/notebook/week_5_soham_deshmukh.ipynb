## Q1. Key Limitations of MapReduce vs Spark

**Question:** What are the key limitations of traditional MapReduce that make Spark a preferred choice for modern big data processing?

### Answer:
MapReduce is a framework used to process large amounts of data in distributed systems.

**Key Limitations:**
- MapReduce writes the output of every step to disk, which makes processing slower
- Not suitable for tasks like machine learning or real-time analytics because these tasks need repeated calculations
- Spark solves this problem by keeping data in memory, making it much faster

**Example:** If we want to train a machine learning model many times, Spark finishes the task much faster because it does not read and write data to disk after every step.

## Q2. In-Memory Computing in Spark

**Question:** Explain how Spark uses In-Memory Computing to speed up iterative machine learning algorithms compared to disk-based systems.

### Answer:
In-Memory Computing means Spark stores data in RAM instead of reading it from disk repeatedly.

**Key Benefits:**
- Many machine learning algorithms process the same dataset multiple times
- Spark keeps the data in memory, so it can reuse it quickly
- This reduces disk access and improves performance

**Example:** If an algorithm runs 20 iterations, Spark loads the data once and reuses it, while MapReduce reads the data from disk every iteration.

## Q3. Remove Duplicate Rows Based on Specific Columns

```python
from pyspark.sql import SparkSession
import os

os.environ['PYSPARK_PYTHON'] = 'python'
os.environ['PYSPARK_DRIVER_PYTHON'] = 'python'

spark = SparkSession.builder.appName("RemoveDuplicates").master("local[*]").getOrCreate()

# Sample synthetic data
data = [
    (101, "2026-07-01", 2500),
    (101, "2026-07-01", 2500),
    (102, "2026-07-02", 1800),
    (103, "2026-07-03", 3200),
    (103, "2026-07-03", 3200)
]

columns = ["user_id", "transaction_date", "amount"]

df = spark.createDataFrame(data, columns)

# Remove duplicate rows based on user_id and transaction_date
df_clean = df.dropDuplicates(["user_id", "transaction_date"])

print("Original Data")
df.show()

print("\nAfter Removing Duplicates")
df_clean.show()
```

**Explanation:** This removes duplicate records that have the same `user_id` and `transaction_date`.

## Q4. Filter and Group by with Aggregation

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import avg
import os

os.environ['PYSPARK_PYTHON'] = 'python'
os.environ['PYSPARK_DRIVER_PYTHON'] = 'python'

spark = SparkSession.builder.appName("SalesAnalysis").master("local[*]").getOrCreate()

# Sample synthetic data
data = [
    ("West", "Electronics", 4500),
    ("West", "Electronics", 5500),
    ("West", "Furniture", 3000),
    ("West", "Furniture", 3400),
    ("East", "Electronics", 6000),
    ("North", "Furniture", 2800)
]

columns = ["region", "product_category", "sale_amount"]

df_sales = spark.createDataFrame(data, columns)

# Filter West region and calculate average sale amount
result = (
    df_sales
    .filter(df_sales.region == "West")
    .groupBy("product_category")
    .agg(avg("sale_amount").alias("Average_Sales"))
)

print("Original Sales Data")
df_sales.show()

print("\nAverage Sales for West Region")
result.show()
```

**Explanation:** This query first filters the records for the West region and then groups them by product category to calculate the average sale amount.

## Q5. Difference Between .na.drop() and .na.fill()

```python
from pyspark.sql import SparkSession
import os

os.environ['PYSPARK_PYTHON'] = 'python'
os.environ['PYSPARK_DRIVER_PYTHON'] = 'python'

spark = SparkSession.builder.appName("FillNullValues").master("local[*]").getOrCreate()

# Sample synthetic data
data = [
    (101, "Active"),
    (102, None),
    (103, "Inactive"),
    (104, None)
]

columns = ["user_id", "status"]

df = spark.createDataFrame(data, columns)

# Replace null values in status column
df_clean = df.na.fill({"status": "Unknown"})

print("Original Data")
df.show()

print("\nAfter Filling Null Values")
df_clean.show()
```

**Key Differences:**
- `.na.drop()` removes rows that contain null values
- `.na.fill()` replaces null values with a specified value

**Explanation:** This replaces all null values in the `status` column with `"Unknown"`.

## Q6. Count Records by City with Filter

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import count
import os

os.environ['PYSPARK_PYTHON'] = 'python'
os.environ['PYSPARK_DRIVER_PYTHON'] = 'python'

spark = SparkSession.builder.appName("CityCount").master("local[*]").getOrCreate()

# Sample synthetic data
data = [
    ("Pune", 120),
    ("Mumbai", 150),
    ("Nashik", 80),
    ("Nagpur", 105)
]

columns = ["city", "records"]

df = spark.createDataFrame(data, columns)

# Find cities having more than 100 records
result = (
    df
    .filter(df.records > 100)
    .select("city", "records")
)

print("Original Data")
df.show()

print("\nCities with More Than 100 Records")
result.show()
```

**Explanation:** This displays only the cities where the total number of records is greater than 100.

## Q7. Immutability of Spark DataFrames

```python
from pyspark.sql import SparkSession
import os

os.environ['PYSPARK_PYTHON'] = 'python'
os.environ['PYSPARK_DRIVER_PYTHON'] = 'python'

spark = SparkSession.builder.appName("Immutability").master("local[*]").getOrCreate()

# Sample synthetic data
data = [
    ("Rahul", 21, "Pune"),
    ("Sneha", 22, "Mumbai")
]

columns = ["name", "age", "city"]

df = spark.createDataFrame(data, columns)

# Create a new DataFrame after dropping age column
new_df = df.drop("age")

print("Original DataFrame")
df.show()

print("\nNew DataFrame (after dropping 'age')")
new_df.show()
```

**Key Concept:** Spark DataFrames are immutable, which means they cannot be modified after they are created. Whenever we perform operations like dropping or renaming columns, Spark creates a new DataFrame instead of changing the existing one.

## Q8. Filter with Multiple Conditions

```python
from pyspark.sql import SparkSession
import os

os.environ['PYSPARK_PYTHON'] = 'python'
os.environ['PYSPARK_DRIVER_PYTHON'] = 'python'

spark = SparkSession.builder.appName("PremiumUsers").master("local[*]").getOrCreate()

# Sample synthetic data
data = [
    ("Rahul", 22, "Premium"),
    ("Sneha", 31, "Premium"),
    ("Amit", 19, "Basic"),
    ("Priya", 28, "Premium")
]

columns = ["name", "age", "subscription"]

df = spark.createDataFrame(data, columns)

# Filter Premium users between age 18 and 30
result = (
    df.filter(
        (df.age >= 18) &
        (df.age <= 30) &
        (df.subscription == "Premium")
    )
)

print("Original Data")
df.show()

print("\nFiltered Data (Premium users, age 18-30)")
result.show()
```

**Explanation:** This filters only the Premium users whose age lies between 18 and 30.

## Q9. Handling Null Values Before Aggregations

**Question:** When cleaning a dataset, why is it often better to handle null values before performing mathematical aggregations like `sum()` or `avg()`?

### Answer:
Null values represent missing information in a dataset.

**Why Handle Nulls First:**
- If null values are not handled before performing calculations, the final result may become incorrect
- Some records may be ignored during aggregation
- Replacing null values with 0 or removing that row before calculating the average gives a more reliable result

**Example:** If the prices are `100`, `200` and `null`, replacing the null value with `0` or removing that row before calculating the average gives a more reliable result.

## Q10. Cast and Rename Timestamp Column

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
from pyspark.sql.types import TimestampType
import os

os.environ['PYSPARK_PYTHON'] = 'python'
os.environ['PYSPARK_DRIVER_PYTHON'] = 'python'

spark = SparkSession.builder.appName("TimestampConversion").master("local[*]").getOrCreate()

# Sample synthetic data
data = [
    ("2026-07-15 10:30:00",),
    ("2026-07-16 03:45:10",)
]

columns = ["raw_timestamp"]

df = spark.createDataFrame(data, columns)

# Cast timestamp and rename column
df_new = (
    df
    .withColumn(
        "raw_timestamp",
        col("raw_timestamp").cast(TimestampType())
    )
    .withColumnRenamed("raw_timestamp", "event_time")
)

print("Original Data")
df.show()

print("\nUpdated Data (cast to TimestampType and renamed)")
df_new.show()
```

**Explanation:** This converts the `raw_timestamp` column into `TimestampType` and renames it as `event_time`.

## Q11. Shuffle Process in Grouping Operations

**Question:** Explain the "Shuffle" process that occurs during a grouping operation. Why is it considered a wide transformation?

### Answer:
Shuffle is the process of moving data between different partitions in Spark during operations like `groupBy()` or `join()`.

**Key Points:**
- During a grouping operation, Spark collects similar records from different partitions and brings them together
- Since data is transferred across different partitions in the cluster, it is called a wide transformation
- Shuffle usually takes more time because it involves network communication

**Example:** If sales records of the same city are stored in different partitions, Spark first collects all those records together and then calculates the total sales for that city.

## Q12. Remove Rows with Null or Empty Values

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
import os

os.environ['PYSPARK_PYTHON'] = 'python'
os.environ['PYSPARK_DRIVER_PYTHON'] = 'python'

spark = SparkSession.builder.appName("CleanUserData").master("local[*]").getOrCreate()

# Sample synthetic data
data = [
    ("rahul@gmail.com", "rahul"),
    (None, "amit"),
    ("priya@gmail.com", ""),
    ("neha@gmail.com", "neha")
]

columns = ["email", "username"]

df = spark.createDataFrame(data, columns)

# Remove rows having null email or empty username
clean_df = (
    df.filter(
        col("email").isNotNull() &
        (col("username") != "")
    )
)

print("Original Data")
df.show()

print("\nCleaned Data")
clean_df.show()
```

**Explanation:** This removes all rows where the email is missing or the username is empty.

## Q13. Multiple Aggregations with .agg()

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import min, max, mean
import os

os.environ['PYSPARK_PYTHON'] = 'python'
os.environ['PYSPARK_DRIVER_PYTHON'] = 'python'

spark = SparkSession.builder.appName("AggregateFunctions").master("local[*]").getOrCreate()

# Sample synthetic data
data = [
    (150,),
    (300,),
    (450,),
    (600,),
    (900,)
]

columns = ["price"]

df = spark.createDataFrame(data, columns)

# Calculate multiple statistics
result = (
    df.agg(
        min("price").alias("Minimum"),
        max("price").alias("Maximum"),
        mean("price").alias("Average")
    )
)

print("Original Data")
df.show()

print("\nAggregated Result")
result.show()
```

**Explanation:** This calculates the minimum, maximum and average values of the `price` column in a single query.

## Q14. Risks of inferSchema=true with Inconsistent Data

**Question:** In the context of cleaning a dataset, what is the risk of using `inferSchema=true` when your source data contains messy or inconsistent date formats?

### Answer:
`inferSchema=true` automatically detects the data type of each column while reading a dataset.

**Risks:**
- If the dataset contains inconsistent date formats, Spark may detect the wrong data type
- Some values may be converted into null
- This can create problems during filtering, sorting and analysis

**Example:** If the dataset contains dates like `2026-07-15`, `15/07/2026` and `07-15-2026`, Spark may not correctly identify all of them as date values.

## Q15. Complete Data Cleaning Pipeline

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum
import os

os.environ['PYSPARK_PYTHON'] = 'python'
os.environ['PYSPARK_DRIVER_PYTHON'] = 'python'

spark = SparkSession.builder.appName("FinalPipeline").master("local[*]").getOrCreate()

# Sample synthetic data
data = [
    (101, 1200),
    (101, 1200),
    (102, None),
    (102, 1800),
    (103, 2500)
]

columns = ["store_id", "price"]

df = spark.createDataFrame(data, columns)

# Complete processing pipeline
result = (
    df
    .dropDuplicates()
    .na.fill({"price": 0})
    .groupBy("store_id")
    .agg(sum("price").alias("Total_Revenue"))
)

print("Original Data")
df.show()

print("\nProcessed Data (duplicates removed, nulls filled, grouped by store)")
result.show()
```

**Explanation:** This pipeline:
1. Removes duplicate records
2. Replaces null prices with `0`
3. Calculates the total revenue for each store
