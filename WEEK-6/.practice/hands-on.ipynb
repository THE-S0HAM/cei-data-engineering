```python
import os
import sys

print(sys.executable)

os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable
```

```python
import pyspark as ps

spark = (
    ps.sql.SparkSession.builder
    .master("local[*]")
    .appName("Practice")
    .config("spark.pyspark.python", sys.executable)
    .config("spark.pyspark.driver.python", sys.executable)
    .getOrCreate()
)

sc = spark.sparkContext

print(sc.pythonExec)
```

```python
lines = sc.textFile("file:///c:/Users/Soham Deshmukh/Desktop/B.Tech/Celebal Internship/cei-data-engineering/WEEK-6/.practice/data.txt")
```

```python
#split the lines into words, filter, and count
count= lines.flatMap(lambda x: x.split(" ")) \
    .filter(lambda x: len(x) > 0) \
    .map(lambda x: (x, 1)) \
    .reduceByKey(lambda x, y: x + y)

result = count.collect()
print(result)
```

```python
df = spark.read.csv(
    r"C:\Users\Soham Deshmukh\Desktop\B.Tech\Celebal Internship\cei-data-engineering\WEEK-6\.practice\data.txt",
    header=True,
    inferSchema=True
)
```

```python
df.printSchema()
```

```python
#sql-like transformations
collection= df.select('name', 'department', 'salary') \
            .filter(df.age > 30) \
            .groupBy('department') \
            .count()

collection.show()
```

```python
#sql-like transformations
collection= df.select('name', 'department', 'salary') \
            .filter(df.salary > 80000) \
            .groupBy('department') \
            .count()

collection.show()
```
